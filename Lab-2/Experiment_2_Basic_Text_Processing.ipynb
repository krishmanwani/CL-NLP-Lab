# Experiment-2: Basic Text Preprocessing

## Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:
1. Convert text to lowercase. (Also display how many uppercase?)
2. Remove punctuation.        (Also display how many different types of punctuations?)
3. Remove numbers.            (Also display how many numbers?)
4. Remove extra whitespace. (How much extra whitespace is present?)
5. Display the cleaned text.  


```python
import re
import string

# Load input file
file_path = "2.1_text_data.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# 1. Convert text to lowercase & count uppercase letters
uppercase_count = sum(1 for char in text if char.isupper())
lowercase_text = text.lower()

# 2. Remove punctuation & count different punctuation types
found_punctuations = set(
    char for char in text if char in string.punctuation
)

punctuation_types_count = len(found_punctuations)

no_punctuation_text = lowercase_text.translate(
    str.maketrans("", "", string.punctuation)
)

# 3. Remove numbers & count total digits
digits_count = len(re.findall(r'\d', text))

no_numbers_text = re.sub(
    r'\d+',
    '',
    no_punctuation_text
)

# 4. Remove extra whitespace & count extra spaces
cleaned_text = re.sub(
    r'\s+',
    ' ',
    no_numbers_text
).strip()

extra_whitespace_count = len(no_numbers_text) - len(cleaned_text)

# Display results
print("----- TEXT PREPROCESSING RESULTS -----")

print(f"Uppercase Characters Count: {uppercase_count}")

print(
    f"Distinct Punctuation Types Found: "
    f"{punctuation_types_count} -> "
    f"{', '.join(sorted(found_punctuations))}"
)

print(f"Total Numbers/Digits Found: {digits_count}")

print(f"Extra Whitespace Characters Removed: {extra_whitespace_count}")

print("\n----- CLEANED TEXT -----")
print(cleaned_text)
```

## Experiment 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.
## (Three separate code: using NLTK, using spaCy  and without using NLTK/spacy)


### Tokenization Without NLTK/spaCy

```python
import re

# Load input file
file_path = "2.2_tokenization_data.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text_data = file.read()

# -------------------------------
# Sentence Tokenization
# -------------------------------

sentences = [
    s.strip()
    for s in re.split(r'(?<=[.!?])\s+', text_data)
    if s.strip()
]

# -------------------------------
# Word Tokenization
# -------------------------------

words = re.findall(
    r'\b\w+\b|[^\w\s]',
    text_data
)

# Display sentences
print("----- SENTENCE TOKENIZATION (Without NLTK/spaCy) -----")

for i, sentence in enumerate(sentences, 1):
    print(f"{i}. {sentence}")

# Display words
print("\n----- WORD TOKENIZATION (Without NLTK/spaCy) -----")
print(words)
```

### Tokenization Using NLTK

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

# Download required resources
nltk.download('punkt')
nltk.download('punkt_tab')

# Load input file
file_path = "2.2_tokenization_data.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text_data = file.read()

# -------------------------------
# Tokenization
# -------------------------------

sentences_nltk = sent_tokenize(text_data)
words_nltk = word_tokenize(text_data)

# Display sentences
print("----- SENTENCE TOKENIZATION (NLTK) -----")

for i, sentence in enumerate(sentences_nltk, 1):
    print(f"{i}. {sentence}")

# Display words
print("\n----- WORD TOKENIZATION (NLTK) -----")
print(words_nltk)
```

### Tokenization Using spaCy

```python
import spacy

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Load input file
file_path = "2.2_tokenization_data.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text_data = file.read()

# Process text
doc = nlp(text_data)

# -------------------------------
# Sentence Tokenization
# -------------------------------

sentences_spacy = [
    sentence.text
    for sentence in doc.sents
]

# -------------------------------
# Word Tokenization
# -------------------------------

words_spacy = [
    token.text
    for token in doc
]

# Display sentences
print("----- SENTENCE TOKENIZATION (spaCy) -----")

for i, sentence in enumerate(sentences_spacy, 1):
    print(f"{i}. {sentence}")

# Display words
print("\n----- WORD TOKENIZATION (spaCy) -----")
print(words_spacy)
```

## Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens.
## (Also, display Stop Words Separately).

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download required resources
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

# Load input file
file_path = "2.3_clean_data.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text_data = file.read()

# 1. Tokenize original text
original_tokens = word_tokenize(text_data)

# 2. Get English stop words
stop_words_set = set(stopwords.words('english'))

# 3. Separate stop words and filtered tokens
detected_stop_words = [
    token
    for token in original_tokens
    if token.lower() in stop_words_set
]

filtered_tokens = [
    token
    for token in original_tokens
    if token.lower() not in stop_words_set
]

# -------------------------------
# Display Results
# -------------------------------

print("----- ORIGINAL TOKENS -----")
print(original_tokens)

print(
    f"\n----- STOP WORDS SEPARATELY "
    f"(Total Detected: {len(detected_stop_words)}) -----"
)
print(detected_stop_words)

print("\n----- FILTERED TOKENS (Stop Words Removed) -----")
print(filtered_tokens)
```
