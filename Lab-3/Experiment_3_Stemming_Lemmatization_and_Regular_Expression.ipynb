## Experiment 3.1:  Write a Python program to perform  Write a Python program to perform stemming using the Porter Stemmer. Apply stemming to the following words. Display the original words and their stems.
## playing, played, plays, studies, studying, connected, connection, computers


```python
from nltk.stem import PorterStemmer

porter = PorterStemmer()

words = [
    'playing', 'played', 'plays', 'studies', 'studying',
    'connected', 'connection', 'computers'
]

print("Original word -> Stemmed word")
print("----------------------------")

for word in words:
    stem = porter.stem(word)
    print(f"{word} -> {stem}")
```

## Experiment 3.2:  Write a Python program to perform lemmatization on a given input words:

## cats,	dogs,   running,  runs,	ran, studies, studying,	better, children, mice,	went, ate, leaves,	caring,


```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words_for_lemmatization = [
    'cats', 'dogs', 'running', 'runs', 'ran', 'studies', 'studying',
    'better', 'children', 'mice', 'went', 'ate', 'leaves', 'caring'
]

print("Original word -> Lemmatized word")

for word in words_for_lemmatization:

    lemma_noun = lemmatizer.lemmatize(word, pos='n')

    lemma_verb = lemmatizer.lemmatize(word, pos='v')

    lemma_adj = lemmatizer.lemmatize(word, pos='a')

    if lemma_verb != word and lemma_verb != lemma_noun:
        lemma = lemma_verb
    elif lemma_adj != word and lemma_adj != lemma_noun:
        lemma = lemma_adj
    else:
        lemma = lemma_noun

    print(f"{word} -> {lemma}")
```

```python
import re

text = (
    "Welcome to the Natural Language Processing workshop! For more information, visit "
    "https://www.nlpworkshop.com or www.python.org. You can contact the coordinator at "
    "nlpworkshop@gmail.com or support@python.org. For registration, call +91-9876543210 or 9123456789. "
    "Follow us on social media @NLPWorkshop and @PythonLearner. Share your experience "
    "using #NLP, #Python, and #MachineLearning. You can also visit https://github.com/NLPWorkshop "
    "for the latest updates."
)

# 1. Extract Email addresses
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
emails = re.findall(email_pattern, text)
print("Email Addresses:", emails)

# 2. Extract URLs
# This pattern covers URLs starting with http(s):// and www., as well as those without a scheme but with a domain.
url_pattern = r'\b(?:https?://|www\.)\S+\b|\b[a-zA-Z0-9.-]+\.[a-zA-Z]{2,}(?:/[^\s]*)?'
urls = re.findall(url_pattern, text)
urls = [url for url in urls if '.' in url[url.find('//')+2:] or url.startswith('www.')]
print("URLs:", urls)

# 3. Extract Mobile numbers
# This pattern covers numbers with and without country code, and with/without hyphens.
phone_pattern = r'\b(?:\+\d{1,2}-)?\d{10}\b'
phone_numbers = re.findall(phone_pattern, text)
print("Mobile Numbers:", phone_numbers)

# 4. Extract Hashtags
hashtag_pattern = r'#\w+'
hashtags = re.findall(hashtag_pattern, text)
print("Hashtags:", hashtags)

# 5. Extract Mentions
mention_pattern = r'@\w+'
mentions = re.findall(mention_pattern, text)
print("Mentions:", mentions)
```
