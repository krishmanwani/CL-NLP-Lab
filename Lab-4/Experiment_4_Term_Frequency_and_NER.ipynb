# Experiment-4: Term Frequency and NER

## Experiment 4.1: To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.
- Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
- Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import nltk
import spacy
import csv
import re
from collections import Counter

# Download NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

# --------------------------------------------
# Step 1: Read input file
# --------------------------------------------

file_path = "/content/4.1_4.2_input.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

print("Original Text:")
print(text)

# --------------------------------------------
# Step 2: Convert to lowercase
# --------------------------------------------

text_lower = text.lower()

# --------------------------------------------
# Step 3: Tokenization using NLTK
# --------------------------------------------

tokens = word_tokenize(text_lower)

# Keep only alphabetic words
words = [word for word in tokens if word.isalpha()]

# --------------------------------------------
# Step 4: Calculate term frequency
# --------------------------------------------

frequency = Counter(words)

print("\nTerm Frequencies:")
for term, freq in frequency.items():
    print(term, ":", freq)

# --------------------------------------------
# Step 5: Save frequency in CSV
# --------------------------------------------

csv_file = "/content/4.1_term_frequency.csv"

with open(csv_file, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for term, freq in frequency.items():
        writer.writerow([term, freq])

print("\nFrequency data saved to:")
print(csv_file)

# --------------------------------------------
# Step 6: Top 10 frequent terms
# --------------------------------------------

top_10 = frequency.most_common(10)

print("\nTop 10 Most Frequent Terms:")
print("--------------------------------")

for term, freq in top_10:
    print(f"{term} -> {freq}")


# ============================================
# Named Entity Recognition using spaCy
# ============================================

print("\nNamed Entity Recognition:")
print("--------------------------------")

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

doc = nlp(text)

for entity in doc.ents:
    print(f"{entity.text} -> {entity.label_}")
```

## Experiment 4.2: To perform Term-Frequency Analysis on a given text without using any toolkit.

- Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
- Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import re
import csv

# --------------------------------------------
# Step 1: Read input file
# --------------------------------------------

file_path = "/content/4.1_4.2_input.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

print("Original Text:")
print(text)

# --------------------------------------------
# Step 2: Convert text to lowercase
# --------------------------------------------

text = text.lower()

# --------------------------------------------
# Step 3: Remove punctuation
# --------------------------------------------

text = re.sub(r'[^\w\s]', '', text)

# --------------------------------------------
# Step 4: Tokenize manually
# --------------------------------------------

words = text.split()

print("\nTokens:")
print(words)

# --------------------------------------------
# Step 5: Calculate frequency manually
# --------------------------------------------

frequency = {}

for word in words:

    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# --------------------------------------------
# Step 6: Display frequencies
# --------------------------------------------

print("\nTerm Frequencies:")
print("--------------------------------")

for term, freq in frequency.items():
    print(f"{term} -> {freq}")

# --------------------------------------------
# Step 7: Save to CSV
# --------------------------------------------

csv_file = "/content/4.2_term_frequency.csv"

with open(csv_file, "w", newline="", encoding="utf-8") as file:

    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for term, freq in frequency.items():
        writer.writerow([term, freq])

print("\nFrequency data saved to:")
print(csv_file)

# --------------------------------------------
# Step 8: Find top 10 terms manually
# --------------------------------------------

sorted_frequency = sorted(
    frequency.items(),
    key=lambda x: x[1],
    reverse=True
)

top_10 = sorted_frequency[:10]

print("\nTop 10 Most Frequent Terms:")
print("--------------------------------")

for term, freq in top_10:
    print(f"{term} -> {freq}")
```

## Experiment 4.3: Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:
1. Read multiple text documents.
2. Convert text to lowercase.
3. Remove punctuation.
4. Tokenize the documents.
5. Calculate TF for every term.
6. Calculate DF for every term.
7. Calculate IDF using the logarithmic formula.
8. Calculate TF-IDF for every term in every document.
9. Display Term, TF, DF, IDF, and TF-IDF at each step.
10. Identify the top 10 terms with the highest TF-IDF score for each document.

Given Input

Consider the following three documents:

- Document 1:
Natural language processing is a field of artificial intelligence.

- Document 2:
Natural language processing helps computers understand human language.

- Document 3:
Machine learning is an important part of artificial intelligence.


```python
import re
import math
import csv

# ============================================
# GIVEN INPUT DOCUMENTS
# ============================================

documents = [
    "Natural language processing is a field of artificial intelligence.",

    "Natural language processing helps computers understand human language.",

    "Machine learning is an important part of artificial intelligence."
]

# ============================================
# STEP 1: Convert to lowercase
# ============================================

lowercase_documents = []

for document in documents:

    document = document.lower()

    lowercase_documents.append(document)

print("============================================")
print("STEP 1: LOWERCASE")
print("============================================")

for i, document in enumerate(lowercase_documents, 1):
    print(f"Document {i}:")
    print(document)
    print()


# ============================================
# STEP 2: Remove punctuation
# ============================================

clean_documents = []

for document in lowercase_documents:

    document = re.sub(r'[^\w\s]', '', document)

    clean_documents.append(document)

print("============================================")
print("STEP 2: REMOVE PUNCTUATION")
print("============================================")

for i, document in enumerate(clean_documents, 1):
    print(f"Document {i}:")
    print(document)
    print()


# ============================================
# STEP 3: Tokenization
# ============================================

tokenized_documents = []

for document in clean_documents:

    tokens = document.split()

    tokenized_documents.append(tokens)

print("============================================")
print("STEP 3: TOKENIZATION")
print("============================================")

for i, tokens in enumerate(tokenized_documents, 1):

    print(f"Document {i}:")
    print(tokens)
    print()


# ============================================
# STEP 4: Calculate TF
# ============================================

tf_documents = []

for tokens in tokenized_documents:

    total_words = len(tokens)

    word_count = {}

    # Count each word
    for word in tokens:

        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1

    # Calculate TF
    tf = {}

    for word, count in word_count.items():

        tf[word] = count / total_words

    tf_documents.append(tf)


print("============================================")
print("STEP 4: TERM FREQUENCY (TF)")
print("============================================")

for i, tf in enumerate(tf_documents, 1):

    print(f"\nDocument {i}")

    print("--------------------------------------------")
    print("Term\t\tTF")

    for term, value in tf.items():

        print(f"{term:<15} {value:.4f}")


# ============================================
# STEP 5: Calculate DF
# ============================================

df = {}

for tf in tf_documents:

    for term in tf.keys():

        if term in df:
            df[term] += 1
        else:
            df[term] = 1


print("\n\n============================================")
print("STEP 5: DOCUMENT FREQUENCY (DF)")
print("============================================")

print("Term\t\tDF")
print("--------------------------------------------")

for term in sorted(df):

    print(f"{term:<15} {df[term]}")


# ============================================
# STEP 6: Calculate IDF
# ============================================

N = len(documents)

idf = {}

for term, document_frequency in df.items():

    idf[term] = math.log(N / document_frequency)


print("\n\n============================================")
print("STEP 6: INVERSE DOCUMENT FREQUENCY (IDF)")
print("============================================")

print("Term\t\tDF\tIDF")
print("--------------------------------------------")

for term in sorted(idf):

    print(
        f"{term:<15}"
        f"{df[term]:<8}"
        f"{idf[term]:.4f}"
    )


# ============================================
# STEP 7: Calculate TF-IDF
# ============================================

tfidf_documents = []

for tf in tf_documents:

    tfidf = {}

    for term, tf_value in tf.items():

        tfidf[term] = tf_value * idf[term]

    tfidf_documents.append(tfidf)


print("\n\n============================================")
print("STEP 7: TF-IDF")
print("============================================")

for i, tfidf in enumerate(tfidf_documents, 1):

    print(f"\nDocument {i}")

    print("--------------------------------------------")
    print("Term\t\tTF\tDF\tIDF\tTF-IDF")

    for term in sorted(tfidf):

        print(
            f"{term:<15}"
            f"{tf_documents[i-1][term]:.4f}\t"
            f"{df[term]}\t"
            f"{idf[term]:.4f}\t"
            f"{tfidf[term]:.4f}"
        )


# ============================================
# STEP 8: Top 10 TF-IDF terms for each document
# ============================================

print("\n\n============================================")
print("STEP 8: TOP 10 TF-IDF TERMS")
print("============================================")

for i, tfidf in enumerate(tfidf_documents, 1):

    sorted_tfidf = sorted(
        tfidf.items(),
        key=lambda x: x[1],
        reverse=True
    )

    top_10 = sorted_tfidf[:10]

    print(f"\nDocument {i}")
    print("--------------------------------------------")

    for term, score in top_10:

        print(f"{term:<20} {score:.4f}")


# ============================================
# STEP 9: Save complete TF-IDF result to CSV
# ============================================

csv_file = "/content/4.3_tfidf_results.csv"

with open(csv_file, "w", newline="", encoding="utf-8") as file:

    writer = csv.writer(file)

    writer.writerow([
        "Document",
        "Term",
        "TF",
        "DF",
        "IDF",
        "TF-IDF"
    ])

    for i, tfidf in enumerate(tfidf_documents, 1):

        for term, score in tfidf.items():

            writer.writerow([
                f"Document {i}",
                term,
                round(tf_documents[i-1][term], 4),
                df[term],
                round(idf[term], 4),
                round(score, 4)
            ])

print("\n============================================")
print("CSV FILE CREATED")
print("============================================")

print(csv_file)
```
